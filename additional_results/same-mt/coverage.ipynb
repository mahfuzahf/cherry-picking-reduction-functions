```python
# Imports
import numpy as np
import pandas as pd

t=80
```

```python
vanilla_endpoints = []
cherry_endpoints = []

for i in range(10):
    # Assuming endpoints are stored in a file named 'endpoints_{i}.npy'
    endpoints = np.load(f'vanilla/trial_{i}/final_column.npy')
    vanilla_endpoints.append(endpoints)

for i in range(10):
    # Assuming endpoints are stored in a file named 'endpoints_{i}.npy'
    endpoints = np.load(f'cherry/0.95/trial_{i}/final_column.npy')
    cherry_endpoints.append(endpoints)
```

```python
# get average length of final columns for vanilla and cherry
vanilla_avg_mt = np.mean([len(endpoints) for endpoints in vanilla_endpoints])
cherry_avg_mt = np.mean([len(endpoints) for endpoints in cherry_endpoints])

print(f"Average length of final columns for vanilla: {vanilla_avg_mt}"
      f"\nAverage length of final columns for cherry: {cherry_avg_mt}")
```

```python
running_coverage = []
merger = set()
for item in vanilla_endpoints:
    merger.update(set(item))
    coverage = len(merger) / 65536
    running_coverage.append(coverage)

print(running_coverage)
```

```python
running_coverage = []
merger = set()
for item in cherry_endpoints:
    merger.update(set(item))
    coverage = len(merger) / 65536
    running_coverage.append(coverage)

print(running_coverage)
```

```python
def calculate_m0(N, t, alpha):  
    mttarget = (2 * N) / (t + 2)
    return (alpha * mttarget) / (1 - alpha)

print(calculate_m0(2**32, 5000, 0.8))
```

## Success Probability

- take average m_t to calculate p
- then use this to calculate p_ell

```python
p_cherry = 1 - ((1 - (cherry_avg_mt / 2**16)) ** t)
p_vanilla = 1 - ((1 - (vanilla_avg_mt / 2**16)) ** t)

print(f"Success probability of cherry: {p_cherry}")
print(f"Success probability of vanilla: {p_vanilla}")

def p_ell(p, ell):
    return 1 - ((1 - p) ** ell)

# figure out how many ell it takes to get to 99.97% success probability for both cherry and vanilla
cherry_ell = 0
vanilla_ell = 0


for i in range(1, 6):
    p_cherry_ell = p_ell(p_cherry, i)
    p_vanilla_ell = p_ell(p_vanilla, i)
    print(f"Success probability of cherry with ell={i}: {p_cherry_ell}")
    print(f"Success probability of vanilla with ell={i}: {p_vanilla_ell}")
```
