```python
# Imports
import numpy as np
import pandas as pd
```

```python
vanilla_endpoints = []
cherry_endpoints = []

for i in range(10):
    # Assuming endpoints are stored in a file named 'endpoints_{i}.npy'
    endpoints = np.load(f'vanilla/trial_{i}/final_column.npy')
    vanilla_endpoints.append(endpoints)

for i in range(10):
    # Assuming endpoints are stored in a file named 'endpoints_{i}.npy'
    endpoints = np.load(f'cherry/0.95/trial_{i}/final_column.npy')
    cherry_endpoints.append(endpoints)
```

```python
running_coverage = []
merger = set()
for item in vanilla_endpoints:
    merger.update(set(item))
    coverage = len(merger) / 65536
    running_coverage.append(coverage)

print(running_coverage)
```

```python
running_coverage = []
merger = set()
for item in cherry_endpoints:
    merger.update(set(item))
    coverage = len(merger) / 65536
    running_coverage.append(coverage)

print(running_coverage)
```

```python
def calculate_m0(N, t, alpha):  
    mttarget = (2 * N) / (t + 2)
    return (alpha * mttarget) / (1 - alpha)

print(calculate_m0(2**32, 5000, 0.8))
```
