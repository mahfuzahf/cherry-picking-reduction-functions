```python
from utils import *
from math import e
import matplotlib.pyplot as plt
```

```python
N = 2 ** 16
p = 1 - e ** (-2)
alpha = 0.95
t = 10000
```

```python
result = optimise_kj(N, p, alpha, 5, 1, t=t, maxiter=100)
```

```python
kjs, m_0, final_cost, m_values = result
plt.plot(kjs)
plt.xlabel('j')
plt.ylabel('k_j')
plt.title('Optimised k_j values')
plt.show()
```
