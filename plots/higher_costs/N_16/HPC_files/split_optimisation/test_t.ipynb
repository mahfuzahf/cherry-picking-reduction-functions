## Imports

```python
import pickle
from math import e, log
from utils import *
import matplotlib.pyplot as plt
```

## Constants

```python
Ns = [2 ** 16, 2 ** 17, 2 ** 18, 2 ** 19, 2 ** 20, 2 ** 21]
N_labels = [16, 17, 18, 19, 20, 21]
p = 1 - e ** (-2)
alpha = 0.95
t = 500
```

```python
## t = 1000
results = []
for N in Ns:
    result = optimise_kj(N, p, alpha, 5, 1, t=t)
    results.append(result)

  

```
