# Find the cost of reductuin functions 

## Import

```python
from hashlib import sha256
import mmh3
import numpy as np
import pandas as pd
import time
```

```python
N = 2 ** 40
t = 10000
```

```python
def H_c(x):
    return sha256(x.to_bytes(8, 'little')).digest()  # return bytes directly

def r_c(N, t, y, i, ell=0):
    seed = int(i) + (ell*t)
    return mmh3.hash(y, seed, signed=False) % N
```

```python
## make a list of 1,000,000 points 
points = list(range(1_000_000))

# make dataframe with "sha256" and "reduction" columns, and 10 rows for each trial
df = pd.DataFrame(columns=["trial", "sha256_time", "reduction_time"])
# make trial column go from 1 - 10
df["trial"] = range(1, 11)
for trial in range(10):
    # time sha256 hashing of all points
    start_time = time.time()
    sha256_digests = [H_c(x) for x in points]
    sha256_time = time.time() - start_time
    
    # time reduction function on all digests
    start_time = time.time()
    reductions = [r_c(N, t, digest, i=0) for digest in sha256_digests]
    reduction_time = time.time() - start_time
    
    # record times in dataframe
    df.loc[trial + 1] = [trial + 1, sha256_time, reduction_time]
```

```python
display(df)
# show average time for sha256 and reduction
print("Average sha256 time:", df["sha256_time"].mean())
print("Average reduction time:", df["reduction_time"].mean())

# write ratio of sha256 : reduction times as 1:n
df["ratio"] = df["sha256_time"] / df["reduction_time"]
print("Average ratio (sha256 : reduction): 1:", df["ratio"].mean())
```
